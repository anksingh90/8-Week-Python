```python
import time

def brew_tea(): # a function should only focus on proforming single task for code reusability.
    start_time = time.time()    # task 1 : recording time
    print('Brewing tea..')      # task 2 : printing the output
    time.sleep(1)
    print('Tea is ready!')
    end_time = time.time()      # time can not be used in other function
    print(f'Task time : {end_time-start_time} sec')

brew_tea()
```
