```python
import sqlite3

book_new = [(1,"The Hunger Games"),
            (2,"Harry Potter"),
            (3,"To Kill a Mockingbird"),
            (4,"The Lord of the Rings")
            ]

with sqlite3.connect("books.db") as myconn:
    cursor = myconn.cursor()
    cursor.executemany('INSERT INTO book_name (id, name) VALUES (?, ?)', book_new)
    myconn.commit()

```
