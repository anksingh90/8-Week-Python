Context Manager - used to handle files
Steps followed - Open a file --> File Operations --> Close the file

open() --> a new file is opened

myconn.write --> data --> hold memory buffer.

close() --> file connection - close, then memory buffer writes in the file.

```python
myconn = open('books_1.txt','w')
myconn.write('The Hunger Games\n')
myconn.write('Harry Potter\n')
myconn.write('To Kill a Mockingbird\n')
myconn.write('The Lord of the Rings\n')
myconn.write('The Great Gatsby\n')
myconn.write('The Fault in Our Stars\n')
```

Q : Read data from file - books.txt
Sort it in alphabatical order
Write sorted file it new file catalog.txt

```python
with open('books.txt', 'r') as input_file:
    with open('catalog.txt', 'w') as output_file:
        books = input_file.read().split('\n')
        books.sort()
        output_file.write('\n'.join(books))
```

Clubing the with open function : 

```python
with open('books.txt', 'r') as input_file, open('catalog.txt', 'w') as output_file:
    books = input_file.read().split('\n')
    books.sort()
    output_file.write('\n'.join(books))
```
