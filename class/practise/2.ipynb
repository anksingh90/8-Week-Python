Polymorphism : "many forms" and allows the same method, function or operator to behave differently depending on the object or data

```python
class Car:
    def fuel_type(self):
        return "Petrol/Diesel fuel"

class ElectricCar(Car):
    def fuel_type(self):
        return "Electric Charge"

e_car = ElectricCar('Tata', 'TiagoEV', '50kWh')
print(e_car.fuel_type())

nexon = ElectricCar('Tata', 'Nexon')
print(e_car.fuel_type())
```
